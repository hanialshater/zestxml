## Sparse semantic atoms vs semantic-ID codes (zero-shot Reuters)

Tests whether a **sparse model can be semantic**: ZestXML's bilinear machinery is
representation-agnostic — it needs a sparse basis *shared by the doc and label sides*,
not necessarily a lexical one. We compare, inside the unmodified ZestXML binary:

1. **lexical** — stemmed unigram+bigram tf-idf (standard ZestXML recipe)
2. **+ atoms** — lexical ⊕ flat k-sparse semantic atoms (corpus tf-idf → LSA →
   k-sparse autoencoder; 512 atoms, k=16, weight 0.3), as `#atom<j>` features on
   **both** Xf and Yf (no underscore in the token, so the direct Xf↔Yf matcher fires)
3. **+ RQ codes** — same LSA embeddings quantized semantic-ID style (residual
   k-means, 4×32) through the identical harness — the ablation that isolates the
   *shape* of the basis

Dataset: GZXML-format zero-shot split of Reuters-21578 via NLTK (7,105 train /
3,019 test docs, 25 of 90 labels unseen). See `semantic_atoms/README.md` for the
full writeup; scripts are dataset-agnostic and run unchanged on GZ-Eurlex-4.3K.

### Reference results

| unseen labels (374 docs) | P@1 | P@3 | R@5 | R@10 |
|---|---|---|---|---|
| lexical baseline | 74.87 | 28.79 | 78.25 | 78.25 |
| + RQ codes (semantic-ID shaped) | 76.20 | 31.11 | 84.43 | 84.43 |
| + sparse atoms | **78.07** | **33.16** | **88.79** | **89.59** |

Seen P@1 stays ~94.2–94.6 in all three. The lexical model's unseen recall
saturates (R@5 = R@10): labels it can't lexically match are never surfaced at any
depth — the atoms fix exactly that. RQ codes help but lose to flat overcomplete
atoms on every unseen metric: tree-path codes are coarse per level and
non-compositional, while k-sparse atoms decompose an unseen label into many
*trained*, shared components.
